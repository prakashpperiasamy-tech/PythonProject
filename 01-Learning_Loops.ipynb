```python
from os import remove

#for Loop - Iterator can be any iterable collection like string, List , Tuple or set
print("For loop - List")
fruits = ["apple", "banana", "cherry"]  #List
for fruit in fruits:
    fruit_upper = fruit.upper()
    print(fruit_upper)
    fruit_lower = fruit.lower()
    print(fruit_lower)

print("\n For loop - Tuple")
colours = ("red", "green", "blue")
for colour in colours:
    print(colour)

print("\n For loop - Set")
countries = {"india", "Nepal", "Srilanka"}
for country in countries:
    print(country)

print("\n For loop - String")
str = "india"
for ch in str:
    print(ch)
```

```python
#List enumerate - returns Tuple - Unpacked in to index and fruit below
fruits = ["apple", "banana", "cherry"]
for index, fruit in enumerate(fruits):
    print(index, fruit)

print("\nPractical Example - Enumerate with 'start' parameter")
Products = ["phone", "Laptop", "computer"]
for index, product in enumerate(Products,start=1):
    print(f"{index} {product:12s}--> Done")     #string padding
```

```python
#Range  and Step in range

print("Assigning Range of values to a list")
alist = list(range(10))
print(alist)

print("to reverse using step")
clist = list(range(10,0,-1))
print(clist)

print("Increasing by 2 in step")
blist = list(range(0,10,2))  # by step
print(blist)

for i in alist:
    print(i,end=",")

print("\n\nRange in for loop")
for num in range(5):
    print(num,end=";")


```

```python
# Range example
queries=list(range(10))
batch_size =3

for batch in range(len(queries) // batch_size):
    start =batch * batch_size
    stop = start + batch_size
    print(f"Batch {batch+1} - {queries[start:stop]}")

```

```python
# For loop - Accumulation pattern
# Build a result list

test_queries = ["where is my order #3042?",
                "I want to cancel my order",
                "Do you have discount?"]
result =[]

for i, query in enumerate(test_queries,start=1):
    if "cancel" in query:
        agent = "Human Agent;"
    elif "discount" in query:
        agent = "Promoter Agent;"
    elif "where" in query:
        agent = "Order Agent;"
    else:
        agent = "General Agent;"
    print(f"{[i]} {query}")
    result.append(agent)
    print(result)
```

```python
#Break and Continue to control loop
#Continue
odds=[]
print(odds)
for n in range(0,10):
    if n % 2 == 0:
        odds.append(n)
print(f"Even number: {odds}")

odds =[]
print(odds)
for n in range(0,10):
    if n % 2 == 0:
        continue            # if condition met continue by going to the next run of the loop
    odds.append(n)          # if the condition is not met this step gets executed.
print(f"Odd number: {odds}")
```

```python
#Break
odds =[]
print(odds)
for n in range(0,10):
    if n >5:
        print(f"Stopped at {n}")
        break            # if condition met break the execution of loop and come to the next step
    odds.append(n)       # if the condition is met this step will not get executed.
print(f"Number: {odds}")
```

```python
# Continue example - Real time
all_review =[{"review_id": 5001, "rating" : 3 , "title": "it is fine"},
             {"review_id": 5002, "rating" : 5 , "title": "Excellent"},
             {"review_id": 5003, "rating" : 4 , "title": "Really good"},
             {"review_id": 5004, "rating" : 2 , "title": "Disappointed"},
             {"review_id": 5005, "rating" : 5 , "title": "Perfect product"},
             ]
good_reviews =[]
max_to_collect =3

print(f"All Review : \n{all_review}")

print("\nItems in List of dict" )
for dic in all_review:
    print(dic)

for dic in all_review:
    if dic["rating"] < 4:
        continue                #Continue the for loop if condition satisfies
    good_reviews.append(dic)

    if len(good_reviews) >= max_to_collect:
        break                   #Break loop and go to the next statement if condition satisfies

print(f"\nGood Reviews :  \n {good_reviews}")

for r in good_reviews:
    print(f"{r["review_id"]:7d} {"*" * r["rating"]} {r["title"]}")
```

```python
#List comprehension and Dictionary Comprehension
good_reviews =[]

print("Normal flow")
for review in all_review:
    if review["rating"] >= 4:
        good_reviews.append(review)
    else:
        continue
print(f"Good Reviews :  \n {good_reviews}")

print("\nSame as List Comprehension")
good_reviews_list_comp = [review for review in all_review if review["rating"] >=4]
print(f"{good_reviews_list_comp}")
```

```python
#list comprehension
# Populate even number for the given range
even_num=[]

print("Normal Flow")
for i in range(1,6):
    even_num.append(i*2)
print(f"{even_num}")

print("\nSame as List Comprehension")
even_num_list_comp = [i*2 for i in range(1,6)]
print(f"{even_num_list_comp}")
```

```python
#List comprehension -  Real time example
products=[
    {"name" : "Monitor", "price" : 5.00, "stock" : 238},
    {"name" : "Laptop", "price" : 10.00, "stock" : 29},
    {"name" : "Headphone", "price" : 12.00, "stock" : 0}
]

in_stock = [p for p in products if p["stock"] > 0]
print("Instock : ",[i["name"] for i in in_stock])

transformed = [
                (f"{i['name']} : ${i['price']:.2f}")
                 for i in in_stock
              ]

print("\nAfter transforming")
for t in transformed:
    print(t)
```

```python
#Dictionary Comprehension
print("Normal flow")
sq_dict = {}
for i in range(1,6):
    sq_dict[i] = i*2
print(sq_dict)

print("Dictionary Comprehension")
even_dict_comp ={i:i*2 for i in range(1,6)}
print(even_dict_comp)
```
