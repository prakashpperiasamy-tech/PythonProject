```python
#Application expect numeric value in "Customer_id" but it is string. We can validate it using Pydantic
customer_data = {
    "customer_id": "1001a",
    "name": "Ravi Kumar",
    "email": "ravi@example.com"
}

print(customer_data["customer_id"])
print("ID type:", type(customer_data["customer_id"]))
print("ID type:", type(customer_data["customer_id"]).__name__)
```

```python
#Pydantic lets us describe the shape of the data and validate incoming values against it.
#Pydantic model is a class that inherits "BaseModel"

from pydantic import BaseModel

class Customer(BaseModel):
    customer_id: int
    name: str
    email: str

customer = Customer(customer_id=1001, name="Ravi Kumar", email="Ravi@xyz.com")
print(customer.customer_id)
print("ID type:", type(customer.customer_id))
print(customer)
#print(customer.dict()) # Deprecated method - instead use model_dump()
print(customer.model_dump())

print("\nPydantic automatically converted the string to integer")
customer = Customer(customer_id="1001", name="Ravi Kumar", email="Ravi@xyz.com")
print(customer.customer_id)
print("ID type:", type(customer.customer_id))
print(customer)   # Pydantic automatically converted the string to integer
print(customer.model_dump())

print(f"""\nPydantic converts "1001" to the integer 1001 because the conversion is valid. It raises ValidationError when it cannot validate the input. In this case '1001a'""")
customer = Customer(customer_id="1001a", name="Ravi Kumar", email="Ravi@xyz.com")
print(customer.customer_id)
print("ID type:", type(customer.customer_id))
print(f"\n{customer}")   # Pydantic tried to convert the string to integer but could not - Error
```
