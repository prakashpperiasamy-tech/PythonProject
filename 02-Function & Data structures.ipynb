```python
#Function
def add(a,b):
    return a+b

add_result = add(5,6)
print(add_result)
```

```python
# Function
#Calculating discount
def apply_discount(price, discount):
    discount_price = price * discount/100
    final_price = price - discount_price
    return final_price

print(apply_discount(1000,10))
print(apply_discount(1000,5))
print(apply_discount(2000,15))
```

```python
#Function
#power
def power(a,exp):
    return a ** exp

print(power(10,2))
print(power(7,3))
```

```python
#Function with default parameter to avoid error if the param is not passed.
#power
def power(a,exp=2):  # Default parameter for "exp" .
    return a ** exp

print(power(10,2))
print(power(7))      #If default parameter is not set in function, it will through error as the 2nd parameter is not passed.
```

```python
# *args - Accepts multiple positional arguments in to Tuple and pass it to function as 1 parameter
def rand_func(*number):
    print(type(number))
    print(number)

rand_func(1,2,3,4,5)
rand_func(1,3,4)

```

```python
#**kwargs: any number of named settings. Two stars collect named arguments ( name=value ) into a dict.

def show_config(**kwargs):
    model = kwargs.get("model", "Default Model")
    temperature = kwargs.get("temperature", 0.7)

    print("All settings:", kwargs)
    print(f"Model: {model}")
    print(f"Temperature: {temperature}")

show_config(model="Claude", temperature=0.2)
show_config(model="GPT")
```

```python
#Combining *args and **kwargs
def process_request(*args, **kwargs):
    print("Inputs : ",args)
    print("Configuration : ",kwargs)

process_request("Where is my order?","Customer_123","Today is Saturday",model="GPT", temperature=0.7)



```

```python
#Method              Does                                          Works by
#append(x)       Adds one item to the end                            —
#extend(items)   Adds each item of another collection                —
#pop() / pop(i)  Removes and returns the last item (or item i )      position
#remove(x)       Removes the first item equal to x                   value

numbers = [1,2,3,4,5]
print("List   : ",numbers)

numbers.append(2) # Add item to the end.
print("Append : ",numbers)

numbers.extend([100,200]) # add items from another List at the end
print("Extend : ",numbers)

numbers.remove(2) # Remove first matching value.
print("After removal of first occurence: ",numbers)

last_number = numbers.pop() # Remove and return the last item or item at the given index.
print("After removal-Pop- Default : ",numbers)

remove_number_byidx = numbers.pop(2) # Remove and return the last item or item at the given index.
print("Removed number -Pop by index(2): ",remove_number_byidx)
print("After removal - Pop by index(2): ",numbers)

#20261102 - 19:28
```
