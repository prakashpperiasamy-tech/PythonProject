```python
#for Loop
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    fruit_upper = fruit.upper()
    print(fruit_upper)
    fruit_lower = fruit.lower()
    print(fruit_lower)

```

```python
#Enumerate - returns Tuple - Unpacked in to index and fruit below
fruits = ["apple", "banana", "cherry"]
for index, fruit in enumerate(fruits):
    print(index, fruit)
```
