```python
import json
import pprint

# Type hint is optional
print("integer")
age : int=30
print(age)
print(type(age))

#Assigning 10 to the variable with Type hint Bool  - Type hint is optional in Python
age : bool=10
print(age)
print(type(age))

print("Addition")
print(age + 10)

print("subtraction")
print(age - 10)

print("multiplication")
print(age * 10)

print("division")
print(age / 10)  # division always returns float

print("Floor division")
print(age // 10) #floor division (drops remainder)

print("Remainder")
print(age % 10)  # remainder only

print("\nFloat type")
price:float=99.9
print(price)
print(type(price))

print(price * 2)
print(price / 10)  # division always returns float
print(price // 2)
```

```python
#String
name="Prakash"
print(name)

#Forward indexing - Index starts with “0
print("Forward indexing")
for i in range(len(name)):
    print(i)
    print(name[i])

#Slicing full string
print("Slicing in forward indexing")
print(name[0:])
print(name[:])

#Slicing substring
print("Slicing in forward indexing")
print(name[0:4])
print(name[3:4])
```

```python
name="Prakash"
#Reverse indexing - Index starts with negative number of the length of the string
print("reversed string :")
print(name[::-1]) # to reverse the full string

#understanding the below for loop in reverse indexing is very important. "i" in for loop start with 0. 0=-0.
print("Understanding reverse indexing using for loop")
for i in range(len(name)): # "i" starts with 0
    print(-i)
    print(name[-i:])

#Slicing
print("Slicing")
print(name[-1:]) # Last character
print(name[:-1]) # Till just before last char

#Slicing substring
print(name[-1:]) # Last character
print(name[-7:]) # from first character till end
```

```python
#List
# Technically heterogeneous — Python allows mixed types
mixed = [1, "hello", 3.14, True]   # valid, but avoid this

# By convention: keep lists homogeneous
scores  = [85, 92, 78, 95]         # all int
models  = ["gpt-4o", "gpt-3.5","gpt-4","gpt-5","gpt-5.5"]   # all str

#Access List elements and slicing
print("Access elements and slicing")
print(scores[0])    # First item
print(scores[-1])   # Last item
print(scores[1:3])  # Slice - starting from index 1 and before index 3

print(models[-2:])  # Last 2 item
print(models[:2])   # first 2 item

#Modify elements
print(f"\nBefore modifying : {models[1]}")
models[1] = "Prakash"
print(f"After modifying : {models[1]}")

print("\n Reversing the list elements")
print(models[::-1])

```

```python
#List - Adding items(Single, Multiple) using append() and extend()
models.append("RAG")  # Adding 1 item at the last
print(models)

models.pop() # removes last item
print(models)

models.extend(["Revathi","Yogapriyan", "Jayapriyan"]) # Adding multiple items at the last
print(models)
```

```python
#List - Removing elements using remove(), pop(), clear()
print(models)
print("\nRemove items from list using value")
models.remove("gpt-5.5")
print(models)

print("\nRemove last item from list using pop")
models.pop()
print(models)

print("\nRemove all the items from list using clear")
models.clear()
print(models)
```

```python
#Dictionary - Collection of "Key : value"
#Nested Dictionary - Dict contains other Dictionaries and lists

customer = {"cust id": 1,
            "cust name":"Prakash",
            "address" :{"street" : "Thirunagar",
                        "city" : "Chennai",
                        "state" :"Tamilnadu",
                        "pin":600026
                       },
            "order":["Ord-1001","Ord-1002"]
            }
print(f"Printing Full dictionary = {customer}")
print(f"\nPrinting Customer name = {customer["cust name"]}")

print(f"\nPrinting state from the address ={customer["address"]["state"]}")

print(f"\nPrinting all Order id for that customer ={customer["order"]}")  # accessing all data from List

print(f"\nPrinting 2nd Order id for that customer ={customer["order"][1]}")  # accessing data from List using index

# Option 1
for i in range(len(customer["order"])):
    print(f"\nOption 1-Printing all Order ids for that customer ={customer["order"][i]}")

# Option 2
for ordr in customer["order"]:
    print(f"\nOption 2-Printing all Order ids for that customer ={ordr}")

# Option 3
for i, ordr in enumerate(customer["order"]):
    print(f"\nOption 3-Printing Order id {i+1} for that customer ={ordr}")
```

```python
#Dictionary - Collection of "Key : value"
#Common Dictionary Methods
user={"name" : "Prakash",
      "Age" : 35,
      "city" : "Chennai",
      "country" : "India",}    # Note: comma at the end is ignored by python - No error thrown

# To get all the keys in the Dictionary
print(user.keys())
print(type(user.keys()))

# To get all the values in the Dictionary
print(user.values())
print(type(user.values()))

# To get all the items from Dictionary
print(user.items())
print(type(user.items()))

# Return value for key from Dictionary, "None" if key not found
print(user.get("name"))
print(type(user.get("name")))

#Example of "None"
print(user.get("Salary"))
print(type(user.get("Salary")))

#To check whether the key exists in Dictionary
Key_avail = "name" in user
print(Key_avail)
Key_avail = "Salary" in user
print(Key_avail)

#To get the number of Key value pairs in Dictionary
Dict_len = len(user)
print(Dict_len)

```

```python
#Dictionary - Collection of "Key : value"
#List of Dictionary
row1 ={"empid" : 1001,
       "empname": "Prakash",
       "salary" : 10000,
       "address": "Chennai",
       "department" : "IT"}

row2 ={"empid" : 1002,
       "empname": "Revathi",
       "salary" : 10001,
       "address": "Chennai",
       "department" : "Finance"}

employee = [row1,row2]                 # List of Dictionaries

print(employee[0]["salary"],employee[0]["empname"], end=",  " )
print()
print(employee[1]["salary"],employee[1]["empname"], end="; ")
print()

print(employee)

print(len(employee))

```

```python
# Nested Dictionary or List of Dictionaries inside other Dictionary
# run above box code before executing this piece of code
row1 ={"empid" : 1001,
       "empname": "Prakash",
       "salary" : 10000,
       "address": [
                   {"type": "Primary", "street":"Thirunagar","state":"Tamilnadu","pincode" : 600026 },
                   {"type": "Secondary", "street":"Byepass","state":"Tamilnadu","pincode" : 123456 },
                  ],
       "department" : "IT"}

print(row1["address"][0])
print(row1["department"])
employee = [row1,row2]
print("Note: Schema of Row1 and Row2 can be different")
print(employee)
```
