```python

# Type hint is optional
print("integer")
age : int=30
print(age)
print(type(age))

#Assigning 10 to the variable with Type hint Bool  - Type hint is optional in Python
age : bool=10
print(age)
print(type(age))

print("Addition")
print(age + 10)

print("subtraction")
print(age - 10)

print("multiplication")
print(age * 10)

print("division")
print(age / 10)  # division always returns float

print("Floor division")
print(age // 10) #floor division (drops remainder)

print("Remainder")
print(age % 10)  # remainder only

print("\nFloat type")
price:float=99.9
print(price)
print(type(price))

print(price * 2)
print(price / 10)  # division always returns float
print(price // 2)
```

```python
#String
name="Prakash"
print(name)

#Forward indexing - Index starts with “0
print("Forward indexing")
for i in range(len(name)):
    print(i)
    print(name[i])

#Slicing full string
print("Slicing in forward indexing")
print(name[0:])
print(name[:])

#Slicing substring
print("Slicing in forward indexing")
print(name[0:4])
print(name[3:4])
```

```python
name="Prakash"
#Reverse indexing - Index starts with negative number of the length of the string
print("reversed string :")
print(name[::-1]) # to reverse the full string

#understanding the below for loop in reverse indexing is very important. "i" in for loop start with 0. 0=-0.
print("Understanding reverse indexing using for loop")
for i in range(len(name)): # "i" starts with 0
    print(-i)
    print(name[-i:])

#Slicing
print("Slicing")
print(name[-1:]) # Last character
print(name[:-1]) # Till just before last char

#Slicing substring
print(name[-1:]) # Last character
print(name[-7:]) # from first character till end
```

```python
#string is immutable
name = "Prakash"
print(name)
print(len(name))
name[0] = "B" # Trying to replace "P" with "B" - Not allowed in string - Will throw error.
print(name)
```

```python
#Boolean - True or False
a : int = 10
b : int = 5

print(a>b)
print(a<b)
print(a==b)
print(a!=b)

print(type(True))
print(type(False))
```

```python
# Memory location
a :int = 10
print(id(a)) # id() shows the actual memory address Python chose

b = a
print(id(b))

if id(a)==id(b) : print("Both id(a) and id(b) refers the same memory location")
```

```python
#f-Strings in Python - Uses {} - For Dynamic text
#Simple usage
name = "Prakash"
age = 47

msg = f"{name} is {age} years old"
print(msg)

```

```python
#f-Strings in Python - Uses {} - For Dynamic text
#Multiple variables
product = "Laptop"
price = 999.9
qty = 5

Total = qty * price

print(f"{product} X {qty} =${Total}")
```

```python
#f-Strings in Python - Uses {} - For Dynamic text
# Including Expressions
a = 10
b = 15

print(f"sum = {a+b}")
print(f"avg = {a+b/2}")
```

```python
#f-Strings in Python - Uses {} - For Dynamic text
# Formating Numbers
price = 19.23457
print(f"price = {price:.2f}")
print(f"percentage={0.123400:.2%}")
print(f"Number with comma={1234:,}")
```

```python
#f-Strings in Python - Uses {} - For Dynamic text
# Multiline  F string

Topic ="RAG"
User = "Prakash"

# print with one blank line
prompt = f"""
You are an expert assistant.
Generate a detailed explanation about {Topic} for {User}.
"""
print("X"*15)
print(prompt)
print("X"*15)

# print without one blank line by introducing "\".
prompt = f"""\
You are an expert assistant.
Generate a detailed explanation about {Topic} for {User}.\
"""
print("#"*15)
print(prompt)
print("#"*15)
```
