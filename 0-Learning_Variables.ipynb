```python
#String
name="Prakash"
print(name)

#Forward indexing - Index starts with “0
print("Forward indexing")
for i in range(len(name)):
    print(i)
    print(name[i])

#Slicing full string
print("Slicing in forward indexing")
print(name[0:])
print(name[:])

#Slicing substring
print("Slicing in forward indexing")
print(name[0:4])
print(name[3:4])
```

```python
name="Prakash"
#Reverse indexing - Index starts with negative number of the length of the string
print("reversed string :")
print(name[::-1]) # to reverse the full string

#understanding the below for loop in reverse indexing is very important. "i" in for loop start with 0. 0=-0.
print("Understanding reverse indexing using for loop")
for i in range(len(name)): # "i" starts with 0
    print(-i)
    print(name[-i:])

#Slicing
print("Slicing")
print(name[-1:]) # Last character
print(name[:-1]) # Till just before last char

#Slicing substring
print(name[-1:]) # Last character
print(name[-7:]) # from first character till end
```

```python
#Dictionary - Collection of "Key : value"
#Nested Dictionary - Dict contains other Dictionaries and lists

customer = {"cust id": 1,
            "cust name":"Prakash",
            "address" :{"street" : "Thirunagar",
                        "city" : "Chennai",
                        "state" :"Tamilnadu",
                        "pin":600026
                       },
            "order":["Ord-1001","Ord-1002"]
            }
print(f"Printing Full dictionary = {customer}")
print(f"\nPrinting Customer name = {customer["cust name"]}")

print(f"\nPrinting state from the address ={customer["address"]["state"]}")

print(f"\nPrinting all Order id for that customer ={customer["order"]}")  # accessing all data from List

print(f"\nPrinting 2nd Order id for that customer ={customer["order"][1]}")  # accessing data from List using index

# Option 1
for i in range(len(customer["order"])):
    print(f"\nOption 1-Printing all Order ids for that customer ={customer["order"][i]}")

# Option 2
for ordr in customer["order"]:
    print(f"\nOption 2-Printing all Order ids for that customer ={ordr}")

# Option 3
for i, ordr in enumerate(customer["order"]):
    print(f"\nOption 3-Printing Order id {i+1} for that customer ={ordr}")
```

```python
#Dictionary - Collection of "Key : value"
#Common Dictionary Methods
user={"name" : "Prakash",
      "Age" : 35,
      "city" : "Chennai",
      "country" : "India",}    # Note: comma at the end is ignored by python - No error thrown

# To get all the keys in the Dictionary
print(user.keys())
print(type(user.keys()))

# To get all the values in the Dictionary
print(user.values())
print(type(user.values()))

# To get all the items from Dictionary
print(user.items())
print(type(user.items()))

# Return value for key from Dictionary, "None" if key not found
print(user.get("name"))
print(type(user.get("name")))

#Example of "None"
print(user.get("Salary"))
print(type(user.get("Salary")))

#To check whether the key exists in Dictionary
Key_avail = "name" in user
print(Key_avail)
Key_avail = "Salary" in user
print(Key_avail)

#To get the number of Key value pairs in Dictionary
Dict_len = len(user)
print(Dict_len)

```

```python
#Set - 20261028
#Create Set - unordered collection of unique dataset - Duplicates removed by default

fruits = {"apple", "banana","apple","cherry","apple",10,True}
print(fruits)  #Unique elements, Different Datatypes and print in unordered format

#Add or remove elements
print("Adding Elements")
fruits.add("orange")
print(fruits)
print("Deleting elements")
fruits.remove("apple")
print(fruits)

#To create empty set use set() - Do not use {}. Using {} will create Dictionary instead of set.
empty_set=set() # Correct
print(f"Items in set :{empty_set}\n Type of empty_set is {type(empty_set)}")

empty_set1={} # InCorrect - This will create Dict instead of set. Do not use {}.
print(f"Items in set1 :{empty_set1}\n Type of empty_set1 is {type(empty_set1)}")
```

```python
#Set - Operations - union, intersection, difference and symmetric difference. Also Subset, superset
a={1,2,3,4,5}
b={5,6,7,3,2}

print("Union Operation :")
print(a|b)
print("Intersection Operation :")
print(a&b)
print("Difference Operation :")
print(a-b)
print("Symmetric difference operation :")
print(a^b)

full_num = a|b
print(f"full number :{full_num}")

basic={3,4,5}
print(f"Basic number :{basic}")

print(basic.issubset(full_num))
print(full_num.issuperset(basic))
```

```python
#Membership operator "in" - Looks for items in List, Tuple and set -  returns True or False
allowed_tool ={"Lookup_order", "Lookup_customer","Check_inventory","Get_Product"}
print(allowed_tool)
print(type(allowed_tool))
print("Check_Inventory" in allowed_tool)  # I in caps hence false
print("Check_inventory" in allowed_tool)  # case matches hence true

requested_tool = "delete_database"

if requested_tool in allowed_tool:
    print(f"calling requested tool {requested_tool} .")
else:
    print(f"Requested tool {requested_tool} is Not Allowed")

Added_tool = allowed_tool.add("pricing")    # Add item to the set.
print(f"Added {Added_tool} to the list.")
print(allowed_tool)

allowed_tool.discard("price") # remove safely- no error if not found
print(f"After discarding : {allowed_tool}")

allowed_tool.remove("price")  # Raise Key error - if not found
print(f"After removing : {allowed_tool}")
```

```python
#Multiple conditions using if, elif and else

customer_query = "Where is my order #3042?"
q=customer_query.lower() # to normalise once

if "cancel" in q or "refund" in q:
    agent, urgency = "Human agent", "High"
elif "track" in q or "where" in q or "delivery" in q:
    agent, urgency = "Order agent", "Normal"
elif "product" in q or "spec" in q:
    agent, urgency = "Catalog agent", "low"
else:
    agent, urgency = "General agent", "Normal"

print(f"query : {customer_query}")
print(f"Agent : {agent}")
print(f"urgency : {urgency}")
```
