```python

# Type hint is optional
print("integer")
age : int=30
print(age)
print(type(age))

#Assigning 10 to the variable with Type hint Bool  - Type hint is optional in Python
age : bool=10
print(age)
print(type(age))

print("Addition")
print(age + 10)

print("subtraction")
print(age - 10)

print("multiplication")
print(age * 10)

print("division")
print(age / 10)  # division always returns float

print("Floor division")
print(age // 10) #floor division (drops remainder)

print("Remainder")
print(age % 10)  # remainder only

print("\nFloat type")
price:float=99.9
print(price)
print(type(price))

print(price * 2)
print(price / 10)  # division always returns float
print(price // 2)
```

```python
#String
name="Prakash"
print(name)

#Forward indexing - Index starts with “0
print("Forward indexing")
for i in range(len(name)):
    print(i)
    print(name[i])

#Slicing full string
print("Slicing in forward indexing")
print(name[0:])
print(name[:])

#Slicing substring
print("Slicing in forward indexing")
print(name[0:4])
print(name[3:4])
```

```python
name="Prakash"
#Reverse indexing - Index starts with negative number of the length of the string
print("reversed string :")
print(name[::-1]) # to reverse the full string

#understanding the below for loop in reverse indexing is very important. "i" in for loop start with 0. 0=-0.
print("Understanding reverse indexing using for loop")
for i in range(len(name)): # "i" starts with 0
    print(-i)
    print(name[-i:])

#Slicing
print("Slicing")
print(name[-1:]) # Last character
print(name[:-1]) # Till just before last char

#Slicing substring
print(name[-1:]) # Last character
print(name[-7:]) # from first character till end
```

```python
#Dictionary - Collection of "Key : value"
# Access keys
user = {"name" : "Prakash",
        "age"  :  35,             # even the comma is used, Dict will not through error
       }

print(user["name"])
print(user["age"])

```

```python
#Dictionary - Collection of "Key : value"
#Add or update values
user = {"name" : "Prakash",
        "age"  :  35
        }

print(f"Existing dictionary = {user}")
print(f"Name = {user["name"]}")
user["name"] = "Revathi"
print(f"Updated Name = {user["name"]}")

user["city"] = "Chennai"
print(f"Added Key value pair = {user["city"]}")
print(f"updated dictionary = {user}")
```

```python
#Dictionary - Collection of "Key : value"
#We can remove Key:value pair using del or pop()
user={"name" : "Prakash",
      "Age" : 35,
      "city" : "Chennai",
      "country" : "India",}    # Note: comma at the end is ignored by python - No error thrown

print(f"Before deleting 'name' = {user}")

del user["name"]
print(f"After deleting 'name' using 'del' = {user}")

user.pop("city")
print(f"After deleting 'city' using 'pop()' = {user}")


```

```python
#Dictionary - Collection of "Key : value"
#Nested Dictionary - Dict contains other Dictionaries and lists

customer = {"cust id": 1,
            "cust name":"Prakash",
            "address" :{"street" : "Thirunagar",
                        "city" : "Chennai",
                        "state" :"Tamilnadu",
                        "pin":600026
                       },
            "order":["Ord-1001","Ord-1002"]
            }
print(f"Printing Full dictionary = {customer}")
print(f"\nPrinting Customer name = {customer["cust name"]}")

print(f"\nPrinting state from the address ={customer["address"]["state"]}")

print(f"\nPrinting all Order id for that customer ={customer["order"]}")  # accessing all data from List

print(f"\nPrinting 2nd Order id for that customer ={customer["order"][1]}")  # accessing data from List using index

# Option 1
for i in range(len(customer["order"])):
    print(f"\nOption 1-Printing all Order ids for that customer ={customer["order"][i]}")

# Option 2
for ordr in customer["order"]:
    print(f"\nOption 2-Printing all Order ids for that customer ={ordr}")

# Option 3
for i, ordr in enumerate(customer["order"]):
    print(f"\nOption 3-Printing Order id {i+1} for that customer ={ordr}")
```

```python
#Dictionary - Collection of "Key : value"
#Common Dictionary Methods
user={"name" : "Prakash",
      "Age" : 35,
      "city" : "Chennai",
      "country" : "India",}    # Note: comma at the end is ignored by python - No error thrown

# To get all the keys in the Dictionary
print(user.keys())
print(type(user.keys()))

# To get all the values in the Dictionary
print(user.values())
print(type(user.values()))

# To get all the items from Dictionary
print(user.items())
print(type(user.items()))

# Return value for key from Dictionary, "None" if key not found
print(user.get("name"))
print(type(user.get("name")))

#Example of "None"
print(user.get("Salary"))
print(type(user.get("Salary")))

#To check whether the key exists in Dictionary
Key_avail = "name" in user
print(Key_avail)
Key_avail = "Salary" in user
print(Key_avail)

#To get the number of Key value pairs in Dictionary
Dict_len = len(user)
print(Dict_len)

```
